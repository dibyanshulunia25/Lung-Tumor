```python
# Block 1: Environment, Reproducibility, Visual Preprocessing & GPU Setup

import torch, torch.nn as nn, torch.optim as optim, torchvision
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader, Subset
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, roc_curve, auc
import matplotlib.pyplot as plt, seaborn as sns, pandas as pd, numpy as np, copy, os, random, time
from google.colab import drive, files

drive.mount('/content/drive')

# GPU Optimization & Verification
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"\n{'='*50}\n🚀 HARDWARE CHECK\n{'='*50}")
print(f"Using Device: {device.type.upper()}")
if device.type == 'cuda':
    print(f"GPU Name: {torch.cuda.get_device_name(0)}")
    torch.backends.cudnn.benchmark = True # Optimizes convolution operations for static input sizes
else:
    print("⚠️ WARNING: GPU not detected. Execution will be slow.")
print('='*50 + "\n")

def set_seed(seed):
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)
    np.random.seed(seed); random.seed(seed)
    torch.backends.cudnn.deterministic = False

def get_transforms(train=False):
    t = [transforms.Resize((224, 224))]
    if train: t.extend([transforms.RandomHorizontalFlip(), transforms.RandomVerticalFlip(), transforms.RandomRotation(90)])
    t.extend([transforms.ToTensor(), transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])])
    return transforms.Compose(t)

def visualize_multi_samples(data_dir, num_samples=5):
    dataset = datasets.ImageFolder(data_dir); class_names = dataset.classes; transform = get_transforms(train=True)
    fig, axes = plt.subplots(len(class_names) * num_samples, 2, figsize=(10, 3 * len(class_names) * num_samples))
    row = 0
    for class_idx, class_name in enumerate(class_names):
        idxs = [i for i, (p, l) in enumerate(dataset.imgs) if l == class_idx]
        for idx in random.sample(idxs, num_samples):
            raw, _ = dataset[idx]; pre = transform(raw)
            pre = (pre.permute(1,2,0).numpy() - pre.min().item()) / (pre.max().item() - pre.min().item())
            axes[row, 0].imshow(raw); axes[row, 0].set_title(f"Original: {class_name}"); axes[row, 0].axis('off')
            axes[row, 1].imshow(pre); axes[row, 1].set_title(f"Preprocessed"); axes[row, 1].axis('off')
            row += 1
    plt.tight_layout(); plt.show()

set_seed(42)
data_path = '/content/drive/MyDrive/Mini Project Dataset/Lung (Adenocarcinoma)'
visualize_multi_samples(data_path)


# Block 2: Data Partitioning (70/15/15) & Strict Overlap Check

def prepare_federated_data(num_clients):
    ds = datasets.ImageFolder(data_path); classes = ds.classes
    idx_by_cls = [[] for _ in range(len(classes))]
    for i, (_, l) in enumerate(ds.imgs): idx_by_cls[l].append(i)

    # 15% Global Test Set
    test_idx = []
    for c in range(len(classes)):
        np.random.shuffle(idx_by_cls[c]); split = int(len(idx_by_cls[c]) * 0.15)
        test_idx.extend(idx_by_cls[c][:split]); idx_by_cls[c] = idx_by_cls[c][split:]

    client_loaders = []; train_indices_all = []; val_indices_all = []

    print(f"\n{'='*50}\n📊 DATASET INVENTORY & OVERLAP CHECK\n{'='*50}")

    for i in range(num_clients):
        c_idx = []; specialty = i % len(classes); dist = {}
        for c in range(len(classes)):
            ratio = 0.6 if c == specialty else 0.1
            num = int(len(idx_by_cls[c]) * ratio)
            c_idx.extend(idx_by_cls[c][:num]); dist[classes[c]] = num
            idx_by_cls[c] = idx_by_cls[c][num:]

        np.random.shuffle(c_idx)
        # 15% Global Val means roughly 17.6% of the remaining 85% data
        v_split = int(len(c_idx) * 0.176)
        c_val, c_train = c_idx[:v_split], c_idx[v_split:]

        val_indices_all.extend(c_val)
        train_indices_all.extend(c_train)

        # Added pin_memory and num_workers for speed
        client_loaders.append({
            'train': DataLoader(Subset(datasets.ImageFolder(data_path, get_transforms(True)), c_train), batch_size=32, shuffle=True, pin_memory=True, num_workers=2),
            'val': DataLoader(Subset(datasets.ImageFolder(data_path, get_transforms(False)), c_val), batch_size=32, pin_memory=True, num_workers=2)
        })
        print(f"🏥 HOSPITAL {i+1} [Focus: {classes[specialty]}] | Total Local: {len(c_idx)}")

    # Strict Overlap Verification
    set_train = set(train_indices_all)
    set_val = set(val_indices_all)
    set_test = set(test_idx)

    overlap_tv = len(set_train.intersection(set_val))
    overlap_tt = len(set_train.intersection(set_test))
    overlap_vt = len(set_val.intersection(set_test))

    total_data = len(set_train) + len(set_val) + len(set_test)

    print(f"\n✅ DATASET SPLIT (Target ~ 70/15/15)")
    print(f"Train: {len(set_train)} ({(len(set_train)/total_data)*100:.1f}%) | Val: {len(set_val)} ({(len(set_val)/total_data)*100:.1f}%) | Test: {len(set_test)} ({(len(set_test)/total_data)*100:.1f}%)")
    print(f"⚠️ OVERLAP CHECKS: Train/Val: {overlap_tv} | Train/Test: {overlap_tt} | Val/Test: {overlap_vt}\n")

    g_test_loader = DataLoader(Subset(datasets.ImageFolder(data_path, get_transforms(False)), test_idx), batch_size=64, pin_memory=True, num_workers=2)
    g_val_loader = DataLoader(Subset(datasets.ImageFolder(data_path, get_transforms(False)), val_indices_all), batch_size=64, pin_memory=True, num_workers=2)
    return client_loaders, g_val_loader, g_test_loader, classes, test_idx

# Block 3: Research Orchestrator & Evaluation Core

def get_model(num_classes):
    model = models.mobilenet_v2(weights=models.MobileNet_V2_Weights.IMAGENET1K_V1)
    for p in model.parameters(): p.requires_grad = False
    for p in model.features[-3:].parameters(): p.requires_grad = True
    model.classifier[1] = nn.Linear(model.last_channel, num_classes)
    return model

def evaluate_full(model, loader, device, return_cm=True):
    model.eval(); y_t, y_p, y_probs = [], [], []; total_loss = 0; crit = nn.CrossEntropyLoss()
    with torch.no_grad():
        for imgs, labels in loader:
            imgs, labels = imgs.to(device), labels.to(device); out = model(imgs)
            total_loss += crit(out, labels).item(); y_t.extend(labels.cpu().numpy())
            y_p.extend(torch.argmax(out, 1).cpu().numpy()); y_probs.extend(torch.softmax(out, 1).cpu().numpy())

    y_t, y_p = np.array(y_t), np.array(y_p)
    metrics = {
        "acc": (y_p == y_t).mean() * 100,
        "f1": f1_score(y_t, y_p, average='weighted', zero_division=0),
        "p": precision_score(y_t, y_p, average='weighted', zero_division=0),
        "r": recall_score(y_t, y_p, average='weighted'),
        "auc": roc_auc_score(y_t, y_probs, multi_class='ovr'),
        "loss": total_loss/len(loader)
    }
    if return_cm:
        metrics["cm"] = confusion_matrix(y_t, y_p)
    return metrics, y_t, y_p

def plot_and_save_roc(y_t, y_probs, classes, conf_name):
    plt.figure(figsize=(8, 6))
    for i in range(len(classes)):
        fpr, tpr, _ = roc_curve(np.array(y_t) == i, np.array(y_probs)[:, i])
        plt.plot(fpr, tpr, label=f'{classes[i]} (AUC={auc(fpr, tpr):.2f})')
    plt.plot([0,1], [0,1], 'k--'); plt.legend(); plt.title(f"ROC Curves - {conf_name}")
    plt.tight_layout(); plt.savefig(f"ROC_{conf_name}.png"); plt.show()

# Block 4: The Main Federated Loop

def run_federated_round(seed, n_clients, n_epochs, clients_list, val_loader, test_loader, class_names):
    set_seed(seed); g_model = get_model(len(class_names)).to(device)
    m_size_mb = sum(p.numel() for p in g_model.parameters()) * 4 / (1024**2)
    best_acc, best_model_wts, patience_counter, patience_limit = 0, copy.deepcopy(g_model.state_dict()), 0, 3

    # Store aggregated training performance across all clients for final report
    final_train_stats = {"acc": [], "p": [], "r": [], "f1": []}

    print(f"\n" + "═"*85 + f"\n⚙️ EXECUTING CONFIG | Seed: {seed} | Clients: {n_clients} | Local Epochs: {n_epochs}\n" + "═"*85)

    for r in range(1, 11): # Kept to 10 rounds to control total execution time, adjust if needed
        round_start_time = time.time()
        active = list(range(n_clients))
        local_ws = []
        # Comm metrics storage for this round
        round_train_acc, round_train_p, round_train_r, round_train_f1 = [], [], [], []

        # Stochastic Absence (10% chance a client drops per round)
        if random.random() > 0.9:
            absent = random.choice(active); active.remove(absent)
            print(f"⚠️ STOCHASTIC ABSENCE: Hospital {absent + 1} is offline.")

        uplink = m_size_mb * len(active); downlink = m_size_mb * len(active)
        print(f"\n🔄 ROUND {r} | Traffic: Up {uplink:.1f}MB / Down {downlink:.1f}MB")

        for i in active:
            l_model = copy.deepcopy(g_model).to(device)
            opt = optim.Adam(filter(lambda p: p.requires_grad, l_model.parameters()), lr=1e-4)

            print(f"\n  🏥 Hospital {i+1} Training:")
            print(f"  {'Epoch':<6} | {'Train Loss':<10} | {'Train Acc':<10} | {'Train P':<10} | {'Train R':<10} |  {'Train F1':<9} | {'Time(s)':<8}")
            print(f"  {'-'*80}")

            for epoch in range(n_epochs):
                e_start = time.time()
                l_model.train(); y_t, y_p, total_l = [], [], 0
                for imgs, lbls in clients_list[i]['train']:
                    imgs, lbls = imgs.to(device), lbls.to(device); opt.zero_grad()
                    out = l_model(imgs); loss = nn.CrossEntropyLoss()(out, lbls); loss.backward(); opt.step()
                    total_l += loss.item(); y_t.extend(lbls.cpu().numpy()); y_p.extend(torch.argmax(out, 1).cpu().numpy())

                # Local Train Metrics
                e_acc = (np.array(y_p) == np.array(y_t)).mean() * 100
                e_p = precision_score(y_t, y_p, average='weighted', zero_division=0)
                e_r = recall_score(y_t, y_p, average='weighted')
                e_f = f1_score(y_t, y_p, average='weighted', zero_division=0)
                print(f"  {epoch+1:<6} | {total_l/len(clients_list[i]['train']):<10.4f} | {e_acc:<10.2f} | {e_p:<6.3f} | {e_r:<6.3f} | {e_f:<9.3f} | {time.time()-e_start:<8.2f}")

                if epoch == n_epochs - 1: # Capture last epoch stats
                    round_train_acc.append(e_acc); round_train_p.append(e_p)
                    round_train_r.append(e_r); round_train_f1.append(e_f)

            local_ws.append({k: v.cpu() for k, v in l_model.state_dict().items()}); del l_model; torch.cuda.empty_cache()

        # Aggregation
        new_glob = {k: torch.stack([w[k].float() for w in local_ws], 0).mean(0) for k in g_model.state_dict().keys()}
        g_model.load_state_dict(new_glob)

        # Global Validation
        val_m, _, _ = evaluate_full(g_model, val_loader, device, return_cm=False)
        print(f"\n  🌐 ROUND {r} AGGREGATION | Val Acc: {val_m['acc']:.2f}% | Val AUC: {val_m['auc']:.3f} | Time: {time.time()-round_start_time:.1f}s")

        if val_m['acc'] > best_acc:
            best_acc, best_model_wts, patience_counter = val_m['acc'], copy.deepcopy(g_model.state_dict()), 0
            # Update best training performance seen so far
            final_train_stats = {"acc": np.mean(round_train_acc), "p": np.mean(round_train_p), "r": np.mean(round_train_r), "f1": np.mean(round_train_f1)}
            print(f"  🌟 New Best Global Model Saved!")
        else:
            patience_counter += 1
            if patience_counter >= patience_limit:
                print(f"  🛑 Early Stopping triggered."); break

    # Final Evaluation for this Config
    g_model.load_state_dict(best_model_wts)

    print("\n" + "─"*50 + "\n📈 GENERATING FINAL TEST METRICS\n" + "─"*50)
    test_m, y_test_t, y_test_p = evaluate_full(g_model, test_loader, device, return_cm=True)

    # Calculate global probabilities for ROC plotting
    g_model.eval(); y_probs = []
    with torch.no_grad():
        for imgs, _ in test_loader: y_probs.extend(torch.softmax(g_model(imgs.to(device)), 1).cpu().numpy())

    conf_name = f"S{seed}_E{n_epochs}"
    plot_and_save_roc(y_test_t, y_probs, class_names, conf_name)

    print(f"📊 TEST ACCURACY: {test_m['acc']:.2f}% | PRECISION: {test_m['p']:.3f} | RECALL: {test_m['r']:.3f} | F1: {test_m['f1']:.3f}")
    print(f"🧩 TEST CONFUSION MATRIX:\n{test_m['cm']}")

    return final_train_stats, test_m, g_model, conf_name

# Block 5: Master Execution & Reporting Hub

configs = [(42, 3), (42, 5), (123, 3), (123, 5)]
c_list, v_l, t_l, class_names, global_test_indices = prepare_federated_data(num_clients=5)

master_summary = []

for seed, epochs in configs:
    train_metrics, test_metrics, trained_model, run_id = run_federated_round(seed, 5, epochs, c_list, v_l, t_l, class_names)

    # Save Model
    model_path = f"FedAvg_Model_{run_id}.pt"
    torch.save(trained_model.state_dict(), model_path)

    # Compile Run Report
    run_data = {
        "Config": run_id, "Seed": seed, "Epochs": epochs,
        "Train_Acc": train_metrics['acc'], "Train_P": train_metrics['p'],
        "Train_R": train_metrics['r'], "Train_F1": train_metrics['f1'],
        "Test_Acc": test_metrics['acc'], "Test_P": test_metrics['p'], "Test_R": test_metrics['r'],
        "Test_F1": test_metrics['f1'], "Test_AUC": test_metrics['auc']
    }
    master_summary.append(run_data)

    # Export specific run
    run_df = pd.DataFrame([run_data])
    excel_path = f"Metrics_{run_id}.xlsx"
    run_df.to_excel(excel_path, index=False)

    # Download files (Note: Browsers may block multiple popups. Allow them in Colab)
    try:
        files.download(excel_path)
        files.download(model_path)
        files.download(f"ROC_{run_id}.png")
    except Exception as e:
        print(f"⚠️ Could not auto-download files for {run_id}. Files are saved in Colab folder.")

# Generate Master Summary Ledger
print("\n" + "🔥"*25 + "\n🏆 ALL CONFIGURATIONS COMPLETE. MASTER SUMMARY:\n" + "🔥"*25)
summary_df = pd.DataFrame(master_summary)
print(summary_df.to_markdown(index=False))

master_excel = "Final_Master_Summary.xlsx"
summary_df.to_excel(master_excel, index=False)
try:
    files.download(master_excel)
except:
    pass
print("\n✅ PIPELINE FINISHED SUCCESSFULLY.")
```

```python
# Block 1: Environment, Reproducibility, Visual Preprocessing & GPU Setup

import torch, torch.nn as nn, torch.optim as optim, torchvision
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader, Subset
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, roc_curve, auc
import matplotlib.pyplot as plt, seaborn as sns, pandas as pd, numpy as np, copy, os, random, time
from google.colab import drive, files

drive.mount('/content/drive')

# GPU Optimization & Verification
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"\n{'='*50}\n🚀 HARDWARE CHECK\n{'='*50}")
print(f"Using Device: {device.type.upper()}")
if device.type == 'cuda':
    print(f"GPU Name: {torch.cuda.get_device_name(0)}")
    torch.backends.cudnn.benchmark = True # Optimizes convolution operations for static input sizes
else:
    print("⚠️ WARNING: GPU not detected. Execution will be slow.")
print('='*50 + "\n")

def set_seed(seed):
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)
    np.random.seed(seed); random.seed(seed)
    torch.backends.cudnn.deterministic = False

def get_transforms(train=False):
    t = [transforms.Resize((224, 224))]
    if train: t.extend([transforms.RandomHorizontalFlip(), transforms.RandomVerticalFlip(), transforms.RandomRotation(90)])
    t.extend([transforms.ToTensor(), transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])])
    return transforms.Compose(t)

def visualize_multi_samples(data_dir, num_samples=5):
    dataset = datasets.ImageFolder(data_dir); class_names = dataset.classes; transform = get_transforms(train=True)
    fig, axes = plt.subplots(len(class_names) * num_samples, 2, figsize=(10, 3 * len(class_names) * num_samples))
    row = 0
    for class_idx, class_name in enumerate(class_names):
        idxs = [i for i, (p, l) in enumerate(dataset.imgs) if l == class_idx]
        for idx in random.sample(idxs, num_samples):
            raw, _ = dataset[idx]; pre = transform(raw)
            pre = (pre.permute(1,2,0).numpy() - pre.min().item()) / (pre.max().item() - pre.min().item())
            axes[row, 0].imshow(raw); axes[row, 0].set_title(f"Original: {class_name}"); axes[row, 0].axis('off')
            axes[row, 1].imshow(pre); axes[row, 1].set_title(f"Preprocessed"); axes[row, 1].axis('off')
            row += 1
    plt.tight_layout(); plt.show()

set_seed(42)
data_path = '/content/drive/MyDrive/Mini Project Dataset/Lung (Adenocarcinoma)'
visualize_multi_samples(data_path)


# Block 2: Data Partitioning (70/15/15) & Strict Overlap Check

def prepare_federated_data(num_clients):
    ds = datasets.ImageFolder(data_path); classes = ds.classes
    idx_by_cls = [[] for _ in range(len(classes))]
    for i, (_, l) in enumerate(ds.imgs): idx_by_cls[l].append(i)

    # 15% Global Test Set
    test_idx = []
    for c in range(len(classes)):
        np.random.shuffle(idx_by_cls[c]); split = int(len(idx_by_cls[c]) * 0.15)
        test_idx.extend(idx_by_cls[c][:split]); idx_by_cls[c] = idx_by_cls[c][split:]

    client_loaders = []; train_indices_all = []; val_indices_all = []

    print(f"\n{'='*50}\n📊 DATASET INVENTORY & OVERLAP CHECK\n{'='*50}")

    for i in range(num_clients):
        c_idx = []; specialty = i % len(classes); dist = {}
        for c in range(len(classes)):
            ratio = 0.6 if c == specialty else 0.1
            num = int(len(idx_by_cls[c]) * ratio)
            c_idx.extend(idx_by_cls[c][:num]); dist[classes[c]] = num
            idx_by_cls[c] = idx_by_cls[c][num:]

        np.random.shuffle(c_idx)
        # 15% Global Val means roughly 17.6% of the remaining 85% data
        v_split = int(len(c_idx) * 0.176)
        c_val, c_train = c_idx[:v_split], c_idx[v_split:]

        val_indices_all.extend(c_val)
        train_indices_all.extend(c_train)

        # Added pin_memory and num_workers for speed
        client_loaders.append({
            'train': DataLoader(Subset(datasets.ImageFolder(data_path, get_transforms(True)), c_train), batch_size=32, shuffle=True, pin_memory=True, num_workers=2),
            'val': DataLoader(Subset(datasets.ImageFolder(data_path, get_transforms(False)), c_val), batch_size=32, pin_memory=True, num_workers=2)
        })
        print(f"🏥 HOSPITAL {i+1} [Focus: {classes[specialty]}] | Total Local: {len(c_idx)}")

    # Strict Overlap Verification
    set_train = set(train_indices_all)
    set_val = set(val_indices_all)
    set_test = set(test_idx)

    overlap_tv = len(set_train.intersection(set_val))
    overlap_tt = len(set_train.intersection(set_test))
    overlap_vt = len(set_val.intersection(set_test))

    total_data = len(set_train) + len(set_val) + len(set_test)

    print(f"\n✅ DATASET SPLIT (Target ~ 70/15/15)")
    print(f"Train: {len(set_train)} ({(len(set_train)/total_data)*100:.1f}%) | Val: {len(set_val)} ({(len(set_val)/total_data)*100:.1f}%) | Test: {len(set_test)} ({(len(set_test)/total_data)*100:.1f}%)")
    print(f"⚠️ OVERLAP CHECKS: Train/Val: {overlap_tv} | Train/Test: {overlap_tt} | Val/Test: {overlap_vt}\n")

    g_test_loader = DataLoader(Subset(datasets.ImageFolder(data_path, get_transforms(False)), test_idx), batch_size=64, pin_memory=True, num_workers=2)
    g_val_loader = DataLoader(Subset(datasets.ImageFolder(data_path, get_transforms(False)), val_indices_all), batch_size=64, pin_memory=True, num_workers=2)
    return client_loaders, g_val_loader, g_test_loader, classes, test_idx

# Block 3: Research Orchestrator & Evaluation Core

def get_model(num_classes):
    model = models.mobilenet_v2(weights=models.MobileNet_V2_Weights.IMAGENET1K_V1)
    for p in model.parameters(): p.requires_grad = False
    for p in model.features[-3:].parameters(): p.requires_grad = True
    model.classifier[1] = nn.Linear(model.last_channel, num_classes)
    return model

def evaluate_full(model, loader, device, return_cm=True):
    model.eval(); y_t, y_p, y_probs = [], [], []; total_loss = 0; crit = nn.CrossEntropyLoss()
    with torch.no_grad():
        for imgs, labels in loader:
            imgs, labels = imgs.to(device), labels.to(device); out = model(imgs)
            total_loss += crit(out, labels).item(); y_t.extend(labels.cpu().numpy())
            y_p.extend(torch.argmax(out, 1).cpu().numpy()); y_probs.extend(torch.softmax(out, 1).cpu().numpy())

    y_t, y_p = np.array(y_t), np.array(y_p)
    metrics = {
        "acc": (y_p == y_t).mean() * 100,
        "f1": f1_score(y_t, y_p, average='weighted', zero_division=0),
        "p": precision_score(y_t, y_p, average='weighted', zero_division=0),
        "r": recall_score(y_t, y_p, average='weighted'),
        "auc": roc_auc_score(y_t, y_probs, multi_class='ovr'),
        "loss": total_loss/len(loader)
    }
    if return_cm:
        metrics["cm"] = confusion_matrix(y_t, y_p)
    return metrics, y_t, y_p

def plot_and_save_roc(y_t, y_probs, classes, conf_name):
    plt.figure(figsize=(8, 6))
    for i in range(len(classes)):
        fpr, tpr, _ = roc_curve(np.array(y_t) == i, np.array(y_probs)[:, i])
        plt.plot(fpr, tpr, label=f'{classes[i]} (AUC={auc(fpr, tpr):.2f})')
    plt.plot([0,1], [0,1], 'k--'); plt.legend(); plt.title(f"ROC Curves - {conf_name}")
    plt.tight_layout(); plt.savefig(f"ROC_{conf_name}.png"); plt.show()

# Block 4: The Main Federated Loop

def run_federated_round(seed, n_clients, n_epochs, clients_list, val_loader, test_loader, class_names):
    set_seed(seed); g_model = get_model(len(class_names)).to(device)
    m_size_mb = sum(p.numel() for p in g_model.parameters()) * 4 / (1024**2)
    best_acc, best_model_wts, patience_counter, patience_limit = 0, copy.deepcopy(g_model.state_dict()), 0, 3

    # Store aggregated training performance across all clients for final report
    final_train_stats = {"acc": [], "p": [], "r": [], "f1": []}

    print(f"\n" + "═"*85 + f"\n⚙️ EXECUTING CONFIG | Seed: {seed} | Clients: {n_clients} | Local Epochs: {n_epochs}\n" + "═"*85)

    for r in range(1, 11): # Kept to 10 rounds to control total execution time, adjust if needed
        round_start_time = time.time()
        active = list(range(n_clients))
        local_ws = []
        # Comm metrics storage for this round
        round_train_acc, round_train_p, round_train_r, round_train_f1 = [], [], [], []

        # Stochastic Absence (10% chance a client drops per round)
        if random.random() > 0.9:
            absent = random.choice(active); active.remove(absent)
            print(f"⚠️ STOCHASTIC ABSENCE: Hospital {absent + 1} is offline.")

        uplink = m_size_mb * len(active); downlink = m_size_mb * len(active)
        print(f"\n🔄 ROUND {r} | Traffic: Up {uplink:.1f}MB / Down {downlink:.1f}MB")

        for i in active:
            l_model = copy.deepcopy(g_model).to(device)
            opt = optim.Adam(filter(lambda p: p.requires_grad, l_model.parameters()), lr=1e-4)

            print(f"\n  🏥 Hospital {i+1} Training:")
            print(f"  {'Epoch':<6} | {'Train Loss':<10} | {'Train Acc':<10} | {'Train P':<10} | {'Train R':<10} |  {'Train F1':<9} | {'Time(s)':<8}")
            print(f"  {'-'*80}")

            for epoch in range(n_epochs):
                e_start = time.time()
                l_model.train(); y_t, y_p, total_l = [], [], 0
                for imgs, lbls in clients_list[i]['train']:
                    imgs, lbls = imgs.to(device), lbls.to(device); opt.zero_grad()
                    out = l_model(imgs); loss = nn.CrossEntropyLoss()(out, lbls); loss.backward(); opt.step()
                    total_l += loss.item(); y_t.extend(lbls.cpu().numpy()); y_p.extend(torch.argmax(out, 1).cpu().numpy())

                # Local Train Metrics
                e_acc = (np.array(y_p) == np.array(y_t)).mean() * 100
                e_p = precision_score(y_t, y_p, average='weighted', zero_division=0)
                e_r = recall_score(y_t, y_p, average='weighted')
                e_f = f1_score(y_t, y_p, average='weighted', zero_division=0)
                print(f"  {epoch+1:<6} | {total_l/len(clients_list[i]['train']):<10.4f} | {e_acc:<10.2f} | {e_p:<6.3f} | {e_r:<6.3f} | {e_f:<9.3f} | {time.time()-e_start:<8.2f}")

                if epoch == n_epochs - 1: # Capture last epoch stats
                    round_train_acc.append(e_acc); round_train_p.append(e_p)
                    round_train_r.append(e_r); round_train_f1.append(e_f)

            local_ws.append({k: v.cpu() for k, v in l_model.state_dict().items()}); del l_model; torch.cuda.empty_cache()

        # Aggregation
        new_glob = {k: torch.stack([w[k].float() for w in local_ws], 0).mean(0) for k in g_model.state_dict().keys()}
        g_model.load_state_dict(new_glob)

        # Global Validation
        val_m, _, _ = evaluate_full(g_model, val_loader, device, return_cm=False)
        print(f"\n  🌐 ROUND {r} AGGREGATION | Val Acc: {val_m['acc']:.2f}% | Val AUC: {val_m['auc']:.3f} | Time: {time.time()-round_start_time:.1f}s")

        if val_m['acc'] > best_acc:
            best_acc, best_model_wts, patience_counter = val_m['acc'], copy.deepcopy(g_model.state_dict()), 0
            # Update best training performance seen so far
            final_train_stats = {"acc": np.mean(round_train_acc), "p": np.mean(round_train_p), "r": np.mean(round_train_r), "f1": np.mean(round_train_f1)}
            print(f"  🌟 New Best Global Model Saved!")
        else:
            patience_counter += 1
            if patience_counter >= patience_limit:
                print(f"  🛑 Early Stopping triggered."); break

    # Final Evaluation for this Config
    g_model.load_state_dict(best_model_wts)

    print("\n" + "─"*50 + "\n📈 GENERATING FINAL TEST METRICS\n" + "─"*50)
    test_m, y_test_t, y_test_p = evaluate_full(g_model, test_loader, device, return_cm=True)

    # Calculate global probabilities for ROC plotting
    g_model.eval(); y_probs = []
    with torch.no_grad():
        for imgs, _ in test_loader: y_probs.extend(torch.softmax(g_model(imgs.to(device)), 1).cpu().numpy())

    conf_name = f"S{seed}_E{n_epochs}"
    plot_and_save_roc(y_test_t, y_probs, class_names, conf_name)

    print(f"📊 TEST ACCURACY: {test_m['acc']:.2f}% | PRECISION: {test_m['p']:.3f} | RECALL: {test_m['r']:.3f} | F1: {test_m['f1']:.3f}")
    print(f"🧩 TEST CONFUSION MATRIX:\n{test_m['cm']}")

    return final_train_stats, test_m, g_model, conf_name

# Block 5: Master Execution & Reporting Hub

configs = [(123, 5)]
c_list, v_l, t_l, class_names, global_test_indices = prepare_federated_data(num_clients=5)

master_summary = []

for seed, epochs in configs:
    train_metrics, test_metrics, trained_model, run_id = run_federated_round(seed, 5, epochs, c_list, v_l, t_l, class_names)

    # Save Model
    model_path = f"FedAvg_Model_{run_id}.pt"
    torch.save(trained_model.state_dict(), model_path)

    # Compile Run Report
    run_data = {
        "Config": run_id, "Seed": seed, "Epochs": epochs,
        "Train_Acc": train_metrics['acc'], "Train_P": train_metrics['p'],
        "Train_R": train_metrics['r'], "Train_F1": train_metrics['f1'],
        "Test_Acc": test_metrics['acc'], "Test_P": test_metrics['p'], "Test_R": test_metrics['r'],
        "Test_F1": test_metrics['f1'], "Test_AUC": test_metrics['auc']
    }
    master_summary.append(run_data)

    # Export specific run
    run_df = pd.DataFrame([run_data])
    excel_path = f"Metrics_{run_id}.xlsx"
    run_df.to_excel(excel_path, index=False)

    # Download files (Note: Browsers may block multiple popups. Allow them in Colab)
    try:
        files.download(excel_path)
        files.download(model_path)
        files.download(f"ROC_{run_id}.png")
    except Exception as e:
        print(f"⚠️ Could not auto-download files for {run_id}. Files are saved in Colab folder.")

# Generate Master Summary Ledger
print("\n" + "🔥"*25 + "\n🏆 ALL CONFIGURATIONS COMPLETE. MASTER SUMMARY:\n" + "🔥"*25)
summary_df = pd.DataFrame(master_summary)
print(summary_df.to_markdown(index=False))

master_excel = "Final_Master_Summary.xlsx"
summary_df.to_excel(master_excel, index=False)
try:
    files.download(master_excel)
except:
    pass
print("\n✅ PIPELINE FINISHED SUCCESSFULLY.")
```

### Visualizing Client Dropout

Here are three alternative visualizations to represent client dropout and its impact on the federated learning process. For these examples, I've created a sample `client_participation` dataset where `1` means a client is active and `0` means a client is absent in a given round.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import random

# Sample data for client participation (1 = active, 0 = absent)
# This is a hypothetical representation, you would use your actual log data.
num_rounds = 10
num_clients = 5

# Let's create a DataFrame to store client participation.
# For demonstration, we'll randomly make some clients absent each round.
client_participation = pd.DataFrame(
    np.ones((num_rounds, num_clients)),
    columns=[f'Hospital {i+1}' for i in range(num_clients)],
    index=[f'Round {r+1}' for r in range(num_rounds)]
)

# Introduce some random absences (1-2 clients per round, or no clients)
for r in range(num_rounds):
    if random.random() < 0.7: # 70% chance of having at least one absence
        num_absent = random.randint(1, min(num_clients, 2)) # 1 or 2 clients absent
        absent_clients = random.sample(range(num_clients), num_absent)
        for client_idx in absent_clients:
            client_participation.iloc[r, client_idx] = 0

# Sample global accuracy data (from previous examples or hypothetical)
# This would be the 'accuracy' list from your `run_federated_round` function
global_accuracy = [
    95.67, 97.01, 96.62, 96.95, 97.20,
    97.55, 97.80, 97.65, 97.90, 98.15
]

print("Sample Client Participation (1=Active, 0=Absent):")
display(client_participation)
```
